```python
from sklearn.linear_model import LogisticRegression
```

```python
from sklearn.model_selection import train_test_split
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, roc_curve, auc
```

```python
features = ['driver_age','driver_gender','driver_race','search_conducted','drugs_related_stop']
```

```python
if set(features + ["is_arrested"]).issubset(df.columns):
    model_df = df[features + ['is_arrested']].dropna()
    print(f"Model Data Frame Created.Size:" ,model_df.shape)
else:
    raise Exception("Missing required columns for modeling")
```

```python
model_df=pd.get_dummies(model_df, columns = ['driver_gender','driver_race'], drop_first = True)
bool_cols = ['search_conducted','drugs_related_stop']
for col in bool_cols:
    if model_df[col].dtype == bool:
        model_df[col] = model_df[col].astype(int)
print("Data is Prepared for Modeling")
```

```python
x=model_df.drop("is_arrested", axis = 1)
y=model_df["is_arrested"]
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size = 0.3, random_state = 42)
```

```python
print(f"training test sizes:{x_train.shape[0]}rows")
```

```python
print(f"training test sizes:{x_test.shape[0]}rows")
```

```python
try:
    clf = LogisticRegression(max_iter = 1000)
    clf.fit(x_train, y_train)
    y_pred = clf.predict(x_test)
    accuracy = accuracy_score(y_test, y_pred)
    print(f"Accuracy: {accuracy}")
except Exception as e:
    print("Error training or evaluating the model:", e)
except Exception as e:
    print("Error Training the Model:", e)
```

```python
try:
    clf = LogisticRegression(max_iter = 100)
    clf.fit(x_train, y_train)
    y_pred = clf.predict(x_test)
    accuracy = accuracy_score(y_test, y_pred)
    print(f"Accuracy: {accuracy}")
except Exception as e:
    print("Error training or evaluating the model:", e)
except Exception as e:
    print("Error Training the Model:", e)
```

```python
try:
    clf = LogisticRegression(max_iter = 10000)
    clf.fit(x_train, y_train)
    y_pred = clf.predict(x_test)
    accuracy = accuracy_score(y_test, y_pred)
    print(f"Accuracy: {accuracy}")
except Exception as e:
    print("Error training or evaluating the model:", e)
except Exception as e:
    print("Error Training the Model:", e)
```

```python
conf_matrix=confusion_matrix(y_test, y_pred)
plt.figure(figsize = (6,5))
sns.heatmap(conf_matrix, annot = True, fmt = "d", cmap = "Blues")
plt.title("Confusion Matrix (correct vs incorrect)")
plt.xlabel("What Model Predicted")
plt.ylabel("what Actual happend")
plt.show()
```

```python
y_probs=clf.predict_proba(x_test)[:,1]
fpr, tpr, thresholds = roc_curve(y_test, y_probs)
roc_auc = auc(fpr, tpr)
plt.figure(figsize = (6,5))
plt.plot(fpr, tpr, label = f"ROC curve Score = {roc_auc:.2f}")
plt.plot([0,1], [0,1],"r--")
plt.title("Roc curve)")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.show()
```
